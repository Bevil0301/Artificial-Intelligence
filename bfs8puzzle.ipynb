```python
from types import SimpleNamespace

START = ((1, 2, 3), (4, 0, 6), (7, 5, 8))
GOAL = ((1, 2, 3), (4, 5, 6), (7, 8, 0))
```

```python
def Node(state, parent=None, action=None):
    node = SimpleNamespace()
    node.state = state
    node.parent = parent
    node.action = action
    return node
```

```python
def is_goal(state):
    if state == GOAL:
        return True
    else:
        return False
```

```python
def Queue():
    queue = []
    return queue
```

```python
def enqueue(queue, node):
    queue.append(node)
```

```python
def dequeue(queue):
    if is_empty(queue):
        return None
    node = queue[0]
    del queue[0]
    return node
```

```python
def is_empty(queue):
    if len(queue) == 0:
        return True
    else:
        return False
```

```python
def child_node(problem, parent, action):
    new_state = problem.result(parent.state, action)
    child = Node(
        state=new_state,
        parent=parent,
        action=action
    )
    return child
```

```python
def solution(node):
    path = []
    while node is not None:
        path.append(node.state)
        node = node.parent
    path.reverse()
    return path
```

```python
def Problem(initial_state, goal_condition, actions, result):
    problem = SimpleNamespace()
    problem.initial_state = initial_state
    problem.goal_condition = goal_condition
    problem.actions = actions
    problem.result = result
    problem.is_goal = goal_condition
    return problem
```

```python
def actions(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                break
        if state[r][c] == 0:
            break

    moves = []
    if r > 0:
        moves.append("Lên")
    if r < 2:
        moves.append("Xuống")
    if c > 0:
        moves.append("Trái")
    if c < 2:
        moves.append("Phải")
    return moves
```

```python
def result(state, action):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                break
        if state[r][c] == 0:
            break

    new_r, new_c = r, c
    if action == "Lên":
        new_r -= 1
    elif action == "Xuống":
        new_r += 1
    elif action == "Trái":
        new_c -= 1
    elif action == "Phải":
        new_c += 1

    board = [list(row) for row in state]
    board[r][c] = board[new_r][new_c]
    board[new_r][new_c] = 0
    return tuple(tuple(row) for row in board)
```

```python
def BFS(problem):
    node = Node(problem.initial_state)
    if problem.is_goal(node.state):
        return solution(node)
    frontier = Queue()
    enqueue(frontier, node)
    explored = set()
    while not is_empty(frontier):
        node = dequeue(frontier)
        explored.add(node.state)
        for action in problem.actions(node.state):
            child = child_node(problem, node, action)
            if child.state not in explored and all(
                child.state != item.state for item in frontier
            ):
                if problem.is_goal(child.state):
                    return solution(child)
                enqueue(frontier, child)
    return None
```

```python
problem = Problem(START, is_goal, actions, result)
path = BFS(problem)
if path is None:
    print("Không tìm thấy lời giải.")
else:
    for step, state in enumerate(path):
        print("Bước", step)
        for row in state:
            print(*row)
        print()
    print("Số bước:", len(path) - 1)
```
